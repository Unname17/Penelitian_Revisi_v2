### 1.) Cleaning

```python
import re,string,nltk

def remove_newline(tweet):
    if tweet is not None and isinstance(tweet, str):
        return re.sub(r'\n+', ' ', tweet)
    return ""

def remove_URL(tweet):
  if tweet is not None and isinstance(tweet, str):
    url = re.compile(r'https?://\S+|www\.\S+')
    return url.sub(r'', tweet)
  else:
    return ""

def remove_html(tweet):
  if tweet is not None and isinstance(tweet, str):
    html = re.compile(r'<.*?>')
    return html.sub(r'', tweet)
  else:
    return ""

def remove_emoji(tweet):
  if tweet is not None and isinstance(tweet, str):
    emoji_pattern = re.compile("["
        u"\U0001F600-\U0001F64F" # emoticons
        u"\U0001F300-\U0001F5FF" # symbols & pictographs
        u"\U0001F680-\U0001F6FF" # transport & map symbols
        u"\U0001F700-\U0001F77F" # alchemical symbols
        u"\U0001F780-\U0001F7FF" # Geometric Shape and pictographs
        u"\U0001F800-\U0001F8FF" # Supplemental Arrows-C
        u"\U0001F900-\U0001F9FF" # Supplemental Symbols and pictographs
        u"\U0001FA00-\U0001FA6F" # Chess Symbols
        u"\U0001FA70-\U0001FAFF" # Symbols and pictograohs Extended-A
        u"\U0001F004-\U0001F0CF" # Additional emoticons
        u"\U0001F004-\U0001F0CF" # flags
                               "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', tweet)
  else:
    return ""

def remove_symbols(tweet):
    if tweet is not None and isinstance(tweet, str):
        tweet = re.sub(r'[^a-zA-Z0-9\s]', ' ', tweet)
        return tweet
    else:
        return ""

def remove_numbers(tweet):
  if tweet is not None and isinstance(tweet, str):
    tweet = re.sub(r'\b\d+\b', '', tweet)
    tweet = re.sub(r'\s+', ' ', tweet).strip()
    return tweet
  return ""

def remove_username(tweet):
  if tweet is not None and isinstance(tweet, str):
    tweet = re.sub(r'@[^\s]+', '', tweet)
    tweet = re.sub(r'\brt\b', '', tweet, flags=re.IGNORECASE)
    return tweet
  return ""

def remove_extra_spaces(tweet):
    if tweet is not None and isinstance(tweet, str):
        return re.sub(r'\s+', ' ', tweet).strip()
    return ""


# ====================================================================
# MULAI PENGECEKAN JUMLAH BARIS
# ====================================================================
print(f"0. JUMLAH AWAL DATA       : {len(df)}")

df['cleaning'] = df['text'].apply(lambda x: remove_newline(x))
print(f"1. Setelah hapus newline  : {len(df)}")

df['cleaning'] = df['cleaning'].apply(lambda x: remove_URL(x))
print(f"2. Setelah hapus URL      : {len(df)}")

df['cleaning'] = df['cleaning'].apply(lambda x: remove_username(x))
print(f"3. Setelah hapus username : {len(df)}")

df['cleaning'] = df['cleaning'].apply(lambda x: remove_html(x))
print(f"4. Setelah hapus HTML     : {len(df)}")

df['cleaning'] = df['cleaning'].apply(lambda x: remove_emoji(x))
print(f"5. Setelah hapus emoji    : {len(df)}")

df['cleaning'] = df['cleaning'].apply(lambda x: remove_symbols(x))
print(f"6. Setelah hapus simbol   : {len(df)}")

df['cleaning'] = df['cleaning'].apply(lambda x: remove_numbers(x))
print(f"7. Setelah hapus angka    : {len(df)}")

df['cleaning'] = df['cleaning'].apply(lambda x: remove_extra_spaces(x)) 
print(f"8. Setelah hapus spasi    : {len(df)}")

df['cleaning'] = df['cleaning'].str.strip()
print(f"9. Setelah str.strip()    : {len(df)}")

# --- BAGIAN KRUSIAL (Tempat baris benar-benar dihapus) ---

df = df[df['cleaning'] != '']
print(f"10. Setelah buang baris kosong : {len(df)}")

df = df.dropna(subset=['cleaning'])
print(f"11. Setelah dropna             : {len(df)}")

df = df.drop_duplicates(subset=['cleaning'], keep='first')
print(f"\n---> 12. HASIL AKHIR SETELAH HAPUS DUPLIKAT: {len(df)} <---")


df.head(10)
```

### 3.) Normalisasi

```python
import pandas as pd
import requests
from io import BytesIO

print(f"Memasuki tahap normalisasi. Total baris saat ini: {len(df)}")

def replace_tweet_words(text, kamus_tidak_baku):
    if not isinstance(text, str) or not text.strip():
        return '', [], [], []

    tokens = str(text).split()

    replaced_words = []
    kata_diganti = []
    kata_baku_terganti = []
    kata_tidak_baku_hash = []

    blacklist_nama = {'mutia', 'prof', 'harjo', 'dekan', 'mbak'}

    for token in tokens:
        token_lower = token.lower()

        if len(token_lower) <= 1 or token_lower.isdigit():
            continue

        if token_lower in blacklist_nama:
            continue

        # Cek ke kamus kata baku (yang sekarang sudah berisi gabungan 2 kamus)
        if token_lower in kamus_tidak_baku:
            baku_word = kamus_tidak_baku[token_lower]

            if isinstance(baku_word, str):
                replaced_words.append(baku_word)
                kata_diganti.append(token)
                kata_baku_terganti.append(baku_word)
                kata_tidak_baku_hash.append(hash(token_lower))
            else:
                replaced_words.append(token)
        else:
            replaced_words.append(token)

    replaced_text = ' '.join(replaced_words)
    return replaced_text, kata_baku_terganti, kata_diganti, kata_tidak_baku_hash


# =========================================================================
# 2. PROSES LOADING KAMUS DAN EKSEKUSI DATA X
# =========================================================================

# --- A. Load Kamus 1: Kamus Excel Spesifik Anda ---
url_excel = "https://github.com/analysisdatasentiment/kamus_kata_baku/raw/main/kamuskatabaku.xlsx"
response = requests.get(url_excel)
file_excel = BytesIO(response.content)
kamus_data_1 = pd.read_excel(file_excel)

dict_kamus_1 = dict(zip(
    kamus_data_1['tidak_baku'].astype(str).str.lower(),
    kamus_data_1['kata_baku'].astype(str)
))

# --- B. Load Kamus 2: Kamus Alay Nasalsabila (CSV) ---
# Mengambil versi raw dari file colloquial-indonesian-lexicon.csv
url_csv = "https://raw.githubusercontent.com/nasalsabila/kamus-alay/master/colloquial-indonesian-lexicon.csv"
kamus_data_2 = pd.read_csv(url_csv)

# Repositori ini menggunakan nama kolom 'slang' (kata tidak baku) dan 'formal' (kata baku)
dict_kamus_2 = dict(zip(
    kamus_data_2['slang'].astype(str).str.lower(),
    kamus_data_2['formal'].astype(str)
))

# --- C. Menggabungkan Kedua Kamus ---
# Kita copy kamus alay sebagai basis, lalu kita timpa (update) dengan kamus Excel Anda.
# Cara ini memastikan jika ada kata yang sama di kedua kamus, 
# aturan dari kamus Excel spesifik Anda yang akan diprioritaskan/dimenangkan.
kamus_tidak_baku_dict = dict_kamus_2.copy()
kamus_tidak_baku_dict.update(dict_kamus_1)

print(f"Total kosakata kamus gabungan: {len(kamus_tidak_baku_dict)} kata")

# --- D. Eksekusi ke DataFrame ---
output_series = df['case_folding'].apply(lambda x: replace_tweet_words(x, kamus_tidak_baku_dict))

df['normalisasi'] = output_series.apply(lambda x: x[0])
df['kata_baku'] = output_series.apply(lambda x: x[1])
df['kata_tidak_baku'] = output_series.apply(lambda x: x[2])
df['kata_Tidak_Baku_hash'] = output_series.apply(lambda x: x[3])

df = df[df['normalisasi'].str.strip() != ''].reset_index(drop=True)

print("✓ Tahap Normalisasi & Filter Kosakata Awal Berhasil!")
df_display_norm = df[['text', 'cleaning', 'case_folding', 'normalisasi']]
df_display_norm.head(15)
```

```python
import pandas as pd
import re

# =========================================================================
# 1. KAMUS SINONIM TERKONTROL (BERDASARKAN OUTPUT CLAUDE)
# =========================================================================
synonym_mapping = {
    # a. Kluster Pengajar/Akademisi -> Target Induk: lecturer
    'lecturers': 'lecturer', 'professor': 'lecturer', 'professors': 'lecturer',
    'prof': 'lecturer', 'tutor': 'lecturer', 'tutors': 'lecturer',
    'instructor': 'lecturer', 'instructors': 'lecturer', 'educator': 'lecturer',
    'educators': 'lecturer', 'academic': 'lecturer', 'academics': 'lecturer',
    'academician': 'lecturer', 'faculty': 'lecturer', 'faculties': 'lecturer',
    'teacher': 'lecturer', 'teachers': 'lecturer', 'lecturing': 'lecturer',

    # b. Kluster Finansial/Kesejahteraan -> Target Induk: salary & welfare
    'salaries': 'salary', 'wage': 'salary', 'wages': 'salary',
    'income': 'salary', 'incomes': 'salary', 'pay': 'salary',
    'paid': 'salary', 'paying': 'salary', 'paycheck': 'salary',
    'fee': 'salary', 'honorarium': 'salary', 'honorariums': 'salary',
    'honorary': 'salary', 'honor': 'salary', 'remuneration': 'salary',
    'compensation': 'salary', 'allowance': 'salary', 'allowances': 'salary',
    'stipend': 'salary', 'tukin': 'salary', 'bonus': 'salary',
    'bonuses': 'salary', 'reward': 'salary', 'rewards': 'salary',
    'pension': 'salary', 'pensions': 'salary',
    'prosperity': 'welfare', 'livelihood': 'welfare',

    # c. Kluster Institusi/Tempat Kerja -> Target Induk: institution
    'university': 'institution', 'universities': 'institution', 'college': 'institution',
    'colleges': 'institution', 'campus': 'institution', 'campuses': 'institution',
    'school': 'institution', 'schools': 'institution', 'institutions': 'institution',
    'sd': 'institution','dpr': 'government',

    # d. Kluster Sistem & Status Kepegawaian -> Target Induk: system & status
    'application': 'system', 'app': 'system', 'apps': 'system', 'platform': 'system',
    'systems': 'system', 'sister': 'system', 'siakad': 'system', 'sinta': 'system',
    'bkd': 'system', 'dapodik': 'system', 'feeder': 'system',
    'pppk': 'status', 'pns': 'status', 'asn': 'status', 'civil servant': 'status',
    'civil servants': 'status', 'servant': 'status', 'servants': 'status',
    'non asn': 'status', 'tenure': 'status', 'tenured': 'status'
}

# =========================================================================
# 2. FUNGSI SUBSTITUSI SINONIM
# =========================================================================
def apply_claude_synonyms(text, synonym_dict):
    # Proteksi jika ada nilai kosong
    if pd.isna(text) or not isinstance(text, str) or not text.strip():
        return ""

    # TANGANI FRASA MULTI-KATA DULU (agar tidak rusak saat di-split)
    # Contoh: 'civil servant' diubah menjadi 'status'
    phrases_to_replace = [
        ('civil servant', 'status'),
        ('civil servants', 'status'),
        ('non asn', 'status')
    ]
    for phrase, target in phrases_to_replace:
        text = re.sub(r'\b' + phrase + r'\b', target, text, flags=re.IGNORECASE)

    # Eksekusi penggantian kata tunggal
    words = text.split()
    replaced_words = []

    for word in words:
        word_lower = word.lower()
        if word_lower in synonym_dict:
            replaced_words.append(synonym_dict[word_lower])
        else:
            replaced_words.append(word)

    return ' '.join(replaced_words)

# =========================================================================
# 3. EKSEKUSI PADA DATAFRAME
# =========================================================================
print("Menjalankan tahap substitusi sinonim (Densifikasi Vektor)...")

# Menerapkan fungsi ke kolom translated_text
df['synonym'] = df['translated_text'].apply(lambda x: apply_claude_synonyms(x, synonym_mapping))

print("✓ Tahap Substitusi Sinonim Selesai! Korpus data Anda telah ter-densifikasi dengan aman.")

# =========================================================================
# 4. TAMPILAN UNTUK LAPORAN BAB 4
# =========================================================================
# Menampilkan komparasi 3 tahap terakhir untuk screenshot laporan
df_display_synonym = pd.DataFrame(df)
df_display_synonym.head(10)
```

## Stop removal

```python
# =========================================================================
# CELL 1: TAHAP STOPWORD REMOVAL (ANTI-NEGASI)
# =========================================================================
import pandas as pd
from nltk.corpus import stopwords

# Mendefinisikan daftar stopword bawaan NLTK
stop_words_id = set(stopwords.words('indonesian'))
stop_words_en = set(stopwords.words('english'))

# PERBAIKAN 1: Kata 'belum' dan 'gak' sudah saya HAPUS dari custom_blacklist ini
custom_blacklist = {
    'even', 'also', 'like', 'really', 'get', 'got', 'would', 'could',
    'make', 'made', 'one', 'two', 'go', 'going', 'say', 'said', 'want',
    'know', 'think', 'well', 'much', 'many', 'way', 'something', 'anything',
    'people', 'still', 'though', 'become', 'yes', 'small', 'year', 'state',
    'first', 'good', 'right', 'take', 'must', 'per', 'sir', 'sis', 'lowk',
    'kalau', 'kita', 'untuk', 'saja', 'ke', 'nya', 'yg', 'aja', 
    'kalo', 'buat', 'dekan', 'mbak', 'ibu', 'pak', 'tapi',
    'yang', 'di', 'ini', 'itu', 'juga', 'kan', 'bagaimana', 'dan', 'harus', 'lagi',
    'sudah', 'dari', 'sama', 'bisa', 'memang', 'dulu', 'bu', 'saya', 'ada',
    'iya', 'begitu', 'terima', 'cuma', 'segini', 'karena', 'baik'
}

# PERBAIKAN 2: Membuat daftar kata negasi yang WAJIB DISELAMATKAN
negation_words = {
    # Kata negasi & slang Bahasa Indonesia
    'tidak', 'bukan', 'belum', 'jangan', 'tak', 'gak', 'enggak', 'ga', 'kurang',
    # Kata negasi Bahasa Inggris (jika ada teks yang sudah terjemahan/campuran)
    'no', 'not', 'nor', 'neither', 'never', 'cannot', 'cant', 'dont', 'doesnt',
    'didnt', 'isnt', 'arent', 'wasnt', 'werent', 'hasnt', 'havent', 'hadnt',
    'wouldnt', 'couldnt', 'shouldnt', 'wont', 'aint'
}

# Gabungkan semua stopword terlebih dahulu
all_stopwords = stop_words_id.union(custom_blacklist).union(stop_words_en)

# PERBAIKAN 3: Kurangi daftar stopword dengan kata negasi.
# Logika: Jika kata negasi ada di all_stopwords, maka akan dihapus dari blacklist.
final_blacklist = all_stopwords - negation_words

# Fungsi pembersihan kata
def final_cleaning_strict(text):
    if pd.isna(text):
        return ""
    words = str(text).split()
    filtered = []

    for w in words:
        w = w.lower()
        # FILTER KETAT: Panjang > 1, bukan angka murni, dan tidak ada di blacklist
        if len(w) > 1 and not w.isdigit() and w not in final_blacklist:
            filtered.append(w)

    return " ".join(filtered)

# Menerapkan pembersihan dan membuat kolom baru
df['stopword_removal'] = df['synonym'].apply(final_cleaning_strict)

# Menampilkan dataframe dengan kolom yang diminta
df[['text', 'cleaning', 'case_folding', 'normalisasi', 'synonym', 'stopword_removal']].head(10)
```

## Lemization

```python
# =========================================================================
# CELL 3: TAHAP LEMMATIZATION (BAHASA INGGRIS - FORMAT TOKEN)
# =========================================================================
import nltk
from nltk.stem import WordNetLemmatizer

# Download kamus dasar (WordNet) jika belum ada
nltk.download('wordnet')
nltk.download('omw-1.4')

# Inisialisasi Lemmatizer
lemmatizer = WordNetLemmatizer()

def apply_lemmatization(tokens):
    lemmatized_tokens = []
    for word in tokens:
        # 1. Lemmatize sebagai kata kerja (verb)
        lemma_verb = lemmatizer.lemmatize(word, pos='v')
        
        # 2. Lemmatize hasilnya sebagai kata benda (noun)
        lemma_final = lemmatizer.lemmatize(lemma_verb, pos='n')
        
        lemmatized_tokens.append(lemma_final)
        
    return lemmatized_tokens

# Terapkan Lemmatization (HASILNYA AKAN TETAP BERUPA LIST/TOKEN)
data_df['lemmatized_tokens'] = data_df['tokenize'].apply(apply_lemmatization)

print("✓ Tahap Lemmatization selesai!")
print("✓ Hasil tetap dalam format Token (List) sesuai kebutuhan LDA Gensim.\n")

# Menampilkan perbandingan BERSEBELAHAN agar Anda tidak perlu cek satu per satu
# Anda bisa langsung melihat perbedaan antara token asli vs token lemmatized
df_display = data_df[['tokenize', 'lemmatized_tokens']]
display(df_display.head(10))
```

### Mencari K ideal

```python
df = pd.read_csv('hasil_processing_Data_15.csv')
df.info()
# Cek dimensi data Anda saat ini
print("Jumlah Baris dan Kolom:", df.shape)
```

```python
df[['text', 'cleaning', 'case_folding', 'normalisasi', 'translated_text', 'synonym', 'stopword_removal','tokenize','lemmatized_tokens']].head(10)
```

```python
import ast
import os
import pickle
import numpy as np
import matplotlib.pyplot as plt
from gensim.corpora import Dictionary
from gensim.models import LdaModel, CoherenceModel
from gensim.models.phrases import Phrases, Phraser

def safe_eval_list(text):
    text = str(text).strip()
    if text.startswith('[') and text.endswith(']'):
        try:
            return ast.literal_eval(text)
        except Exception:
            pass
    text_clean = text.replace("'", "").replace('"', "").replace("[", "").replace("]", "").replace(",", "")
    return text_clean.split()

# 1. Ekstrak teks asli TANPA UBAHAN (Untuk corpus lama/SVM)
texts = df['lemmatized_tokens'].apply(safe_eval_list).tolist()

# Membuat Dictionary & Corpus LAMA (Sama persis sebagai pembanding murni)
dictionary_lama = Dictionary(texts)
dictionary_lama.filter_extremes(no_below=3, no_above=0.5)
corpus_lama = [dictionary_lama.doc2bow(text) for text in texts]

# =========================================================================
# -> BLOK TAMBAHAN: ENTITY RECOGNITION (BIGRAM) & EXTENDED STOPWORDS
# =========================================================================
# A. Perluasan Custom Stopwords (Tambahkan terus ampas baru di sini untuk naikkan akurasi)
custom_stopwords = {
    'no', 'not', 'noy', 
    'come', 'look', 'try', 'let', 'already', 'actually', 'whose', 'different',
    'mrs', 'oh', 'regarding', 'considered', 'indeed', 'etc',
    'wow', 'yesterday', 'especially', 'looking', 'called', 'big', 'high'
}

# B. Manual Entity Recognition (Menggabungkan kata spesifik)
def apply_manual_entities(tokens):
    text_joined = " ".join(tokens)
    text_joined = text_joined.replace("sri mulyani", "sri_mulyani")
    text_joined = text_joined.replace("prabowo subianto", "prabowo_subianto")
    text_joined = text_joined.replace("bpjs ketenagakerjaan", "bpjs_ketenagakerjaan")
    text_joined = text_joined.replace("non asn", "non_asn")
    return text_joined.split()

texts_manual_entities = [apply_manual_entities(t) for t in texts]

# C. Otomatisasi Bigram dengan Gensim
phrases = Phrases(texts_manual_entities, min_count=3, threshold=10)
bigram_model = Phraser(phrases)
texts_bigramed = [bigram_model[t] for t in texts_manual_entities]

# D. Hapus Custom Stopwords dari teks yang sudah di-Bigram
texts_lda = [[word for word in text if word.lower() not in custom_stopwords] for text in texts_bigramed]

# Membuat Dictionary & Corpus KHUSUS LDA REVISI TERBARU
dictionary_lda = Dictionary(texts_lda)
# TIPS AKURASI: Jika terlalu banyak kata langka, naikkan no_below ke 5
dictionary_lda.filter_extremes(no_below=3, no_above=0.5)
corpus_lda = [dictionary_lda.doc2bow(text) for text in texts_lda]

# =========================================================================
# -> MENYIMPAN CORPUS & DICTIONARY SECARA PERMANEN (VERSI 3 / REVISI TERBARU)
# =========================================================================
print("Membekukan dan menyimpan Dictionary serta Corpus ke folder 'model_sidang'...")
os.makedirs("model_sidang", exist_ok=True)

# Simpan versi original (jika belum ada)
dictionary_lama.save("model_sidang/dictionary_lama.dict")
with open("model_sidang/corpus_lama.pkl", "wb") as f:
    pickle.dump(corpus_lama, f)

# Simpan VERSI REVISI (V3)
dictionary_lda.save("model_sidang/dictionary_lda_v3_revisi.dict")
with open("model_sidang/corpus_lda_v3_revisi.pkl", "wb") as f:
    pickle.dump(corpus_lda, f)

print("✓ Dictionary dan Corpus (Revisi Terbaru v3) berhasil dibekukan!\n")

# =========================================================================
# 2. FUNGSI MENGHITUNG AVERAGE TOPIC OVERLAP / STABILITY (Jaccard Similarity)
# =========================================================================
def calculate_overlap(model, num_words=15):
    topics = []
    for i in range(model.num_topics):
        words = set([word for word, prob in model.show_topic(i, topn=num_words)])
        topics.append(words)

    overlaps = []
    for i in range(len(topics)):
        for j in range(i + 1, len(topics)):
            intersection = len(topics[i].intersection(topics[j]))
            union = len(topics[i].union(topics[j]))
            if union > 0:
                overlaps.append(intersection / union)
    return np.mean(overlaps) if overlaps else 0

# =========================================================================
# 3. ITERASI PENCARIAN K OPTIMAL
# =========================================================================
min_topics = 2
max_topics = 10
coherence_scores = []
overlap_scores = []

print("Memulai evaluasi Topic Modeling (menggunakan Corpus LDA Revisi Terbaru)...")
for k in range(min_topics, max_topics + 1):
    lda_model = LdaModel(
        corpus=corpus_lda,
        id2word=dictionary_lda,
        num_topics=k,
        random_state=56,
        chunksize=50,
        passes=5,
        alpha=0.5,
        per_word_topics=True
    )

    coherence_model = CoherenceModel(
        model=lda_model, 
        texts=texts_lda, 
        dictionary=dictionary_lda, 
        coherence='c_v'
    )
    c_v = coherence_model.get_coherence()
    coherence_scores.append(c_v)

    overlap = calculate_overlap(lda_model)
    overlap_scores.append(overlap)

    print(f"K={k} | Coherence: {c_v:.4f} | Overlap: {overlap:.4f}")

# =========================================================================
# 4. MENENTUKAN K IDEAL 
# =========================================================================
k_values = list(range(min_topics, max_topics + 1))
coherence_arr = np.array(coherence_scores)
overlap_arr = np.array(overlap_scores)
gap_scores = coherence_arr - overlap_arr

# K-Ideal kita set berdasarkan hasil evaluasi terbaik Anda (sementara diset 6)
best_k = 4
best_idx = k_values.index(best_k)

print(f"\n>>> K IDEAL (Manual) = {best_k} "
      f"(Coherence={coherence_scores[best_idx]:.4f}, "
      f"Overlap={overlap_scores[best_idx]:.4f}, "
      f"Gap={gap_scores[best_idx]:.4f})\n")

print("Rincian gap tiap K:")
for k, c, o, g in zip(k_values, coherence_scores, overlap_scores, gap_scores):
    marker = "  <-- terpilih" if k == best_k else ""
    print(f"K={k} | Coherence={c:.4f} | Overlap={o:.4f} | Gap={g:.4f}{marker}")

# =========================================================================
# 5. PLOT GRAFIK VISUALISASI 
# =========================================================================
x = range(min_topics, max_topics + 1)
plt.figure(figsize=(10, 6))

plt.plot(x, overlap_scores, label='Average Topic Overlap (Stability)', marker='o', color='steelblue')
plt.plot(x, coherence_scores, label='Topic Coherence (c_v)', marker='s', color='sandybrown')

plt.axvline(x=best_k, color='red', linestyle='--', linewidth=2, label=f'Ideal Topic (K={best_k})')

plt.xlabel('Number of Topics (K)')
plt.ylabel('Score')
plt.title('Topic Coherence and Stability Overlap (LDA V3 Revisi Terbaru)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(x)
plt.tight_layout()

os.makedirs("model_sidang", exist_ok=True)
# Menyimpan gambar plot dengan nama revisi terbaru
plt.savefig(f"model_sidang/coherence_overlap_plot_k{best_k}_v3_revisi.png", dpi=200) 
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.feature_extraction.text import TfidfVectorizer, ENGLISH_STOP_WORDS
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split, cross_val_predict, StratifiedKFold
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

import warnings
from sklearn.exceptions import ConvergenceWarning

# Menekan peringatan untuk tampilan output yang lebih bersih
warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", category=ConvergenceWarning)

# =========================================================================
# 1. PERSIAPAN DATA (PENDEKATAN 2 KELAS SENTIMEN)
# =========================================================================
# Filter ketat hanya untuk Positif dan Negatif
df_model = df[df['Sentiment'].isin(['Positif', 'Negatif'])].copy()

# Fitur X (Baseline) dan Target Y (Sentimen)
X = df_model['lemmatized_tokens'] 
y_label = df_model['Sentiment']

print(f"Total dokumen diproses (Biner): {len(df_model)} (Positif/Negatif)")

# =========================================================================
# 2. TERM WEIGHTING (TF-IDF ANTI-NEGASI - UNIGRAM)
# =========================================================================
# Menyusun kamus stopword empiris: Hapus kata negasi dari daftar buang
standard_stopwords = set(ENGLISH_STOP_WORDS)
negation_words = {'not', 'no', 'nor', 'neither', 'never', 'cannot', 'cant', 'dont', 'doesnt'}
custom_stopwords = list(standard_stopwords - negation_words)

# Terapkan custom_stopwords dengan N-gram (1,1) sesuai hasil eksperimen terbaik
tfidf_vectorizer = TfidfVectorizer(
    ngram_range=(1, 1),
    stop_words=custom_stopwords 
)
X_tfidf = tfidf_vectorizer.fit_transform(X)

print("✅ Persiapan data Baseline, ekstraksi TF-IDF (Unigram), dan Setup Model SVM selesai.")
print(f"Dimensi matriks TF-IDF: {X_tfidf.shape}")
```

## GridSearch CV

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split, cross_val_predict, StratifiedKFold, GridSearchCV
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# =========================================================================
# 3.5. HYPERPARAMETER TUNING DENGAN GRIDSEARCHCV (UNTUK TF-IDF)
# =========================================================================
print("Mulai proses Hyperparameter Tuning dengan GridSearchCV (5-Fold CV)...")
print("Data yang digunakan: X_tfidf\n")

# Kumpulan parameter yang akan diuji
param_grid = {
    'C': [0.1, 1, 10, 100, 1000],  
    'kernel': ['linear', 'rbf'], 
    'gamma': ['scale', 'auto'], 
    'class_weight': ['balanced'] 
}

# Parameter probability=True wajib aktif agar bisa memunculkan persentase keyakinan di Streamlit
svm_base = SVC(probability=True, random_state=42, max_iter=-1)
grid_search = GridSearchCV(
    estimator=svm_base, param_grid=param_grid,
    scoring='accuracy', cv=5, n_jobs=-1, verbose=1
)

# Melatih GridSearchCV dengan TF-IDF
grid_search.fit(X_tfidf, y_label)
svm_model = grid_search.best_estimator_

print(f"\n✅ Tuning Selesai! Model diset dengan parameter: {grid_search.best_params_}")

# =========================================================================
# 4. EVALUASI SEMUA SKENARIO SPLITTING & CROSS-VALIDATION
# =========================================================================
evaluation_results = []
all_predictions = {} 

def calculate_metrics(y_true, y_pred, scenario_name):
    acc = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, average='weighted', zero_division=0)
    rec = recall_score(y_true, y_pred, average='weighted', zero_division=0)
    f1 = f1_score(y_true, y_pred, average='weighted', zero_division=0)

    evaluation_results.append({
        'Skenario': scenario_name,
        'Accuracy': acc, 
        'Precision': prec,
        'Recall': rec,
        'F1-Score': f1
    })

print("\nMengeksekusi evaluasi berbagai Skenario Data Splitting...\n")

# --- A. Skenario Hold-out ---
ratios = {'Hold-out 90:10': 0.10, 'Hold-out 80:20': 0.20, 'Hold-out 70:30': 0.30}
for name, test_size in ratios.items():
    X_train, X_test, y_train, y_test = train_test_split(
        X_tfidf, y_label, test_size=test_size, random_state=42, stratify=y_label
    )
    svm_model.fit(X_train, y_train)
    y_pred = svm_model.predict(X_test)
    calculate_metrics(y_test, y_pred, name)
    all_predictions[name] = {'y_true': y_test, 'y_pred': y_pred}

# --- B. Skenario Cross-Validation ---
cv_scenarios = {'Cross-Val 5-Fold': 5, 'Cross-Val 10-Fold': 10}
for name, folds in cv_scenarios.items():
    cv_strategy = StratifiedKFold(n_splits=folds, shuffle=True, random_state=42)
    y_pred_cv = cross_val_predict(svm_model, X_tfidf, y_label, cv=cv_strategy)
    calculate_metrics(y_label, y_pred_cv, name)
    # Kita simpan juga untuk tabel hasil, tapi nanti tidak digambar di confusion matrix

# =========================================================================
# 5. MENENTUKAN SKENARIO TERBAIK & OUTPUT TABEL
# =========================================================================
df_results = pd.DataFrame(evaluation_results)

# Mencari baris dengan nilai Akurasi tertinggi
best_idx = df_results['Accuracy'].idxmax()
best_scenario_name = df_results.loc[best_idx, 'Skenario']
best_accuracy = df_results.loc[best_idx, 'Accuracy']

# Format ulang angka menjadi 4 desimal
for col in ['Accuracy', 'Precision', 'Recall', 'F1-Score']:
    df_results[col] = df_results[col].apply(lambda x: f"{x:.4f}")

print("="*70)
print("HASIL EVALUASI TF-IDF + SVM (BERBAGAI SKENARIO)")
print("="*70)
print(df_results.to_string(index=False))
print("="*70)
print(f"🌟 SKENARIO TERBAIK: {best_scenario_name} (Akurasi: {best_accuracy:.4f})")
print("="*70)

# =========================================================================
# 6. VISUALISASI CONFUSION MATRIX (HANYA UNTUK 3 SKENARIO HOLD-OUT)
# =========================================================================
labels_order = ['Negatif', 'Positif']

# Menggunakan 1 baris dan 3 kolom khusus untuk data hold-out
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
fig.suptitle(f'Confusion Matrix Skenario Hold-out (TF-IDF + SVM Tuned + Unigram)\nSkenario Terbaik: {best_scenario_name}', 
             fontsize=16, fontweight='bold', y=1.05)

# Filter hanya mengambil skenario hold-out saja untuk divisualisasikan
holdout_keys = ['Hold-out 90:10', 'Hold-out 80:20', 'Hold-out 70:30']

for i, scenario_name in enumerate(holdout_keys):
    data = all_predictions[scenario_name]
    cm = confusion_matrix(data['y_true'], data['y_pred'], labels=labels_order)
    
    cmap_color = 'Oranges' if scenario_name == best_scenario_name else 'Blues'
    title_suffix = ' 🏆 (TERBAIK)' if scenario_name == best_scenario_name else ''
    
    sns.heatmap(cm, annot=True, fmt='d', cmap=cmap_color, cbar=False, ax=axes[i],
                xticklabels=labels_order, yticklabels=labels_order,
                annot_kws={"size": 14, "weight": "bold"})
    
    axes[i].set_title(scenario_name + title_suffix, fontsize=13, fontweight='bold', pad=10)
    axes[i].set_xlabel('Label Prediksi', fontsize=11)
    axes[i].set_ylabel('Label Asli', fontsize=11)

plt.tight_layout()
nama_file_cm = 'cm_holdout_scenarios_tfidf.png'
plt.savefig(nama_file_cm, dpi=300, bbox_inches='tight')
print(f"\n[INFO] Gambar Confusion Matrix Hold-out disimpan sebagai '{nama_file_cm}'")
plt.show()

# =========================================================================
# 7. EKSPOR MODEL FINAL & DATA TESTING UNTUK STREAMLIT
# =========================================================================
print("\n[INFO] Menyiapkan proses ekspor model untuk Streamlit...")

test_ratio = 0.10 if '90:10' in best_scenario_name else (0.20 if '80:20' in best_scenario_name else 0.30)

X_train_final, X_test_final, y_train_final, y_test_final = train_test_split(
    X_tfidf, y_label, test_size=test_ratio, random_state=42, stratify=y_label
)

# Ambil teks mentah (raw text) untuk data testing CSV
_, X_test_raw, _, y_test_raw = train_test_split(
    df_model['text'], y_label, test_size=test_ratio, random_state=42, stratify=y_label
)

# Latih ulang model terbaik pada data training terpilih
svm_model.fit(X_train_final, y_train_final)

# Simpan objek model dan vectorizer (.pkl)
joblib.dump(svm_model, 'svm_model_tuned.pkl')
joblib.dump(tfidf_vectorizer, 'tfidf_vectorizer.pkl')

# 3. Menyimpan model LDA (Opsional jika digunakan dalam alur ini)
print("\n3. Menyimpan model LDA...")
try:
    # Melatih ulang LDA final secara definitif agar yang diekspor PASTI K=4
    # (bukan sisa iterasi K=10 dari cell sebelumnya)
    best_k_final = 4
    lda_model_final = LdaModel(
        corpus=corpus_lda,         # Menggunakan corpus yang bersih
        id2word=dictionary_lda,    # Menggunakan dictionary yang bersih
        num_topics=best_k_final,
        random_state=56,
        chunksize=50,
        passes=5,
        alpha=0.5,
        per_word_topics=True
    )
    
    # Menyimpan model dan dictionary yang benar
    lda_model_final.save('lda_model.model')
    dictionary_lda.save('lda_dictionary.dict') 
    
    print(f"   ✅ 'lda_model.model' (K={best_k_final}) berhasil dilatih dan disimpan.")
    print("   ✅ 'lda_dictionary.dict' berhasil disimpan.")
except NameError as e:
    print(f"   ❌ Gagal menyimpan LDA: {e}")
    print("   Catatan: Pastikan variabel 'corpus_lda' dan 'dictionary_lda' sudah dideklarasikan sebelumnya.")

print("\n=== SEMUA PROSES EKSPOR SELESAI ===")

# Simpan data testing ke CSV untuk fitur uji masal di Streamlit
df_export = pd.DataFrame({
    'Teks_Aspirasi': X_test_raw,
    'Sentiment': y_test_raw
})
df_export.to_csv('data_testing_streamlit.csv', index=False)

print("✅ 'svm_model_tuned.pkl' (Model SVM dengan parameter terbaik) berhasil disimpan.")
print("✅ 'tfidf_vectorizer.pkl' (Vectorizer Unigram) berhasil disimpan.")
print("✅ 'data_testing_streamlit.csv' berhasil diekspor.")
print("🎉 Selesai! Semua file siap disalin ke direktori proyek Streamlit Anda.")
```
