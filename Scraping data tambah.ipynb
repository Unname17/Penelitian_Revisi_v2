```python
import pandas as pd
from apify_client import ApifyClient
from dotenv import load_dotenv
import os

# Initialize the ApifyClient with your API token
load_dotenv()

# Ambil token dari environment variable
api_token = os.getenv("APIFY_TOKEN")

# Initialize the ApifyClient dengan token yang aman
client = ApifyClient(api_token)
# Prepare the Actor input
run_input = {
    "filter:blue_verified": False,
    "filter:consumer_video": False,
    "filter:has_engagement": False,
    "filter:hashtags": False,
    "filter:images": False,
    "filter:links": False,
    "filter:media": False,
    "filter:mentions": False,
    "filter:native_video": False,
    "filter:nativeretweets": False,
    "filter:news": False,
    "filter:pro_video": False,
    "filter:quote": False,
    "filter:replies": False,
    "filter:safe": False,
    "filter:spaces": False,
    "filter:twimg": False,
    "filter:videos": False,
    "filter:vine": False,
    "include:nativeretweets": False,
    "lang": "in",
    "maxItems": 8000,
    "searchTerms": [
        "kesejahteraan dosen OR gaji dosen OR dosen non-ASN OR dosen honorer OR dosen kontrak since:2024-05-01_00:00:00_UTC until:2026-05-01_23:59:59_UTC"
    ]
}

# Run the Actor and wait for it to finish
run = client.actor("CJdippxWmn9uRfooo").call(run_input=run_input)

# Fetch and print Actor results from the run's dataset (if there are any)
items = []
for item in client.dataset(run.defaultDatasetId).iterate_items():
    print(item)
    items.append(item)

df = pd.DataFrame(items)
df.to_excel("data_kesejahteraan_dosen_tambahan.xlsx", index=False)
df
```

```python
import os
import pandas as pd
from apify_client import ApifyClient
from dotenv import load_dotenv

# Memuat token
load_dotenv()
api_token = os.getenv("APIFY_TOKEN")
client = ApifyClient(api_token)

# Gunakan Run ID dari proses Anda yang sudah selesai tadi
run_id = "xL6mct6n0pgMcwxme"

print("Mencari data dari Apify...")
run_info = client.run(run_id).get()

if run_info:
    # PERUBAHAN DI SINI: Menggunakan snake_case
    dataset_id = run_info.default_dataset_id
    
    print(f"Dataset ditemukan! ID: {dataset_id}")
    print("Sedang mengunduh data ke dalam memori, mohon tunggu sebentar...")
    
    # Menarik item tanpa melakukan scraping ulang
    items = list(client.dataset(dataset_id).iterate_items())
    print(f"Berhasil mengunduh {len(items)} baris data!")
    
    # Menyimpan ke Excel
    df = pd.DataFrame(items)
    df.to_excel("data_kesejahteraan_dosen_tambahan.xlsx", index=False)
    print("Selesai! File 'data_kesejahteraan_dosen_tambahan.xlsx' berhasil disimpan di folder ini.")
else:
    print("Run ID tidak ditemukan. Pastikan ID sudah benar.")
```
