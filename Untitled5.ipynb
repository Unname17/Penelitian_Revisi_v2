1.) Cleaning

```python
import re,string,nltk

def remove_newline(tweet):
    if tweet is not None and isinstance(tweet, str):
        return re.sub(r'\n+', ' ', tweet)
    return ""

def remove_URL(tweet):
  if tweet is not None and isinstance(tweet, str):
    url = re.compile(r'https?://\S+|www\.\S+')
    return url.sub(r'', tweet)
  else:
    return ""

def remove_html(tweet):
  if tweet is not None and isinstance(tweet, str):
    html = re.compile(r'<.*?>')
    return html.sub(r'', tweet)
  else:
    return ""

def remove_emoji(tweet):
  if tweet is not None and isinstance(tweet, str):
    emoji_pattern = re.compile("["
        u"\U0001F600-\U0001F64F" # emoticons
        u"\U0001F300-\U0001F5FF" # symbols & pictographs
        u"\U0001F680-\U0001F6FF" # transport & map symbols
        u"\U0001F700-\U0001F77F" # alchemical symbols
        u"\U0001F780-\U0001F7FF" # Geometric Shape and pictographs
        u"\U0001F800-\U0001F8FF" # Supplemental Arrows-C
        u"\U0001F900-\U0001F9FF" # Supplemental Symbols and pictographs
        u"\U0001FA00-\U0001FA6F" # Chess Symbols
        u"\U0001FA70-\U0001FAFF" # Symbols and pictograohs Extended-A
        u"\U0001F004-\U0001F0CF" # Additional emoticons
        u"\U0001F004-\U0001F0CF" # flags
                               "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', tweet)
  else:
    return ""

def remove_symbols(tweet):
    if tweet is not None and isinstance(tweet, str):
        tweet = re.sub(r'[^a-zA-Z0-9\s]', ' ', tweet)
        return tweet
    else:
        return ""

def remove_numbers(tweet):
  if tweet is not None and isinstance(tweet, str):
    tweet = re.sub(r'\b\d+\b', '', tweet)
    tweet = re.sub(r'\s+', ' ', tweet).strip()
    return tweet
  return ""

def remove_username(tweet):
  if tweet is not None and isinstance(tweet, str):
    tweet = re.sub(r'@[^\s]+', '', tweet)
    tweet = re.sub(r'\brt\b', '', tweet, flags=re.IGNORECASE)
    return tweet
  return ""

def remove_extra_spaces(tweet):
    if tweet is not None and isinstance(tweet, str):
        return re.sub(r'\s+', ' ', tweet).strip()
    return ""


df['cleaning'] = df['text'].apply(lambda x: remove_newline(x))
df['cleaning'] = df['cleaning'].apply(lambda x: remove_URL(x))
df['cleaning'] = df['cleaning'].apply(lambda x: remove_username(x))
df['cleaning'] = df['cleaning'].apply(lambda x: remove_html(x))
df['cleaning'] = df['cleaning'].apply(lambda x: remove_emoji(x))
df['cleaning'] = df['cleaning'].apply(lambda x: remove_symbols(x))
df['cleaning'] = df['cleaning'].apply(lambda x: remove_numbers(x))
df['cleaning'] = df['cleaning'].apply(lambda x: remove_extra_spaces(x)) 
df['cleaning'] = df['cleaning'].str.strip()
df = df[df['cleaning'] != '']
df = df.dropna(subset=['cleaning'])
df = df.drop_duplicates(subset=['cleaning'],keep='first')
df.head(10)
```

# 3.NORMALISASI

```python
print(f"Memasuki tahap normalisasi. Total baris saat ini: {len(df)}")

def replace_tweet_words(text, kamus_tidak_baku):
    if not isinstance(text, str) or not text.strip():
        return '', [], [], []

    tokens = str(text).split()

    replaced_words = []
    kata_diganti = []
    kata_baku_terganti = []
    kata_tidak_baku_hash = []

    blacklist_nama = {'mutia', 'prof', 'harjo', 'dekan', 'mbak'}

    for token in tokens:
        token_lower = token.lower()

        if len(token_lower) <= 1 or token_lower.isdigit():
            continue

        if token_lower in blacklist_nama:
            continue

        # Cek ke kamus kata baku
        if token_lower in kamus_tidak_baku:
            baku_word = kamus_tidak_baku[token_lower]

            if isinstance(baku_word, str):
                replaced_words.append(baku_word)
                kata_diganti.append(token)
                kata_baku_terganti.append(baku_word)
                kata_tidak_baku_hash.append(hash(token_lower))
            else:
                replaced_words.append(token)
        else:
            replaced_words.append(token)

    replaced_text = ' '.join(replaced_words)
    return replaced_text, kata_baku_terganti, kata_diganti, kata_tidak_baku_hash

# =========================================================================
# 2. PROSES LOADING KAMUS DAN EKSEKUSI DATA X
# =========================================================================
url = "https://github.com/analysisdatasentiment/kamus_kata_baku/raw/main/kamuskatabaku.xlsx"
response = request.get(url)
file_excel = BytesIO(response.content)
kamus_data = pd.read_excel(file_excel)

kamus_tidak_baku_dict = dict(zip(
    kamus_data['tidak_baku'].astype(str).str.lower(),
    kamus_data['kata_baku'].astype(str)
))

output_series = df['case_folding'].apply(lambda x: replace_tweet_words(x, kamus_tidak_baku_dict))

df['normalisasi'] = output_series.apply(lambda x: x[0])
df['kata_baku'] = output_series.apply(lambda x: x[1])
df['kata_tidak_baku'] = output_series.apply(lambda x: x[2])
df['kata_Tidak_Baku_hash'] = output_series.apply(lambda x: x[3])

df = df[df['normalisasi'].str.strip() != ''].reset_index(drop=True)

print("✓ Tahap Normalisasi & Filter Kosakata Awal Berhasil!")
df_display_norm = df[['text', 'cleaning', 'case_folding', 'normalisasi']]
df_display_norm.head(15)
```

#4.Translate

```python
def pre_translate_normalization(text):
    if not isinstance(text, str):
        return ""
    # Mengamankan singkatan gelar dan institusi agar tidak salah terjemah
    text = re.sub(r'\bs1s2\b', 'bachelor and master degree', text, flags=re.IGNORECASE)
    text = re.sub(r'\bs2s3\b', 'master and doctoral degree', text, flags=re.IGNORECASE)
    text = re.sub(r'\bs1\b', 'bachelor degree', text, flags=re.IGNORECASE)
    text = re.sub(r'\bs2\b', 'master degree', text, flags=re.IGNORECASE)
    text = re.sub(r'\bs3\b', 'doctoral degree', text, flags=re.IGNORECASE)
    text = re.sub(r'\bsd\b', 'institution', text, flags=re.IGNORECASE)
    return text

# Menerapkan Lapis 1
df['normalisasi_fixed'] = df['normalisasi'].apply(pre_translate_normalization)

# =========================================================================
# 2. IDENTIFIKASI BAHASA & ISOLASI KALIMAT UNIK
# =========================================================================
df_normalisasi = pd.DataFrame(df['normalisasi_fixed'].astype(str))
unique_elements = df_normalisasi['normalisasi_fixed'].unique()

print(f"Menemukan {len(unique_elements)} kalimat unik dari total {len(df_normalisasi)} baris data.")
print("Memulai proses penerjemahan otomatis ke Bahasa Inggris (CHUNKING & RETRY AKTIF)...\n")

# =========================================================================
# 3. PROSES PENERJEMAHAN MASAL DENGAN CHUNKING & EXPONENTIAL BACKOFF
# =========================================================================
def safe_translate(text, max_retries=3):
    # Memotong teks panjang menjadi maksimal 2000 karakter agar diterima API
    chunks = textwrap.wrap(text, width=2000, break_long_words=False)
    translated_chunks = []

    for chunk in chunks:
        success = False
        wait_time = 3 # Waktu tunggu awal jika kena rate-limit

        for attempt in range(max_retries):
            try:
                time.sleep(0.5) # Jeda standar yang aman
                # Set source ke 'id' agar API tidak bingung membaca slang
                res = GoogleTranslator(source='id', target='en').translate(chunk)
                translated_chunks.append(res)
                success = True
                break
            except Exception as e:
                print(f"    -> API Sibuk/Error. Menunggu {wait_time} detik... (Percobaan {attempt+1}/{max_retries})")
                time.sleep(wait_time)
                wait_time *= 2 # Exponential backoff (3s -> 6s -> 12s)

        if not success:
            translated_chunks.append(chunk) # Jika tetap gagal, kembalikan teks aslinya

    return " ".join(translated_chunks)

translations = {}

# Looping ke seluruh kalimat unik
for i, element in enumerate(unique_elements):
    text_strip = str(element).strip()

    if not text_strip:
        translations[element] = ""
        continue

    # Indikator progres
    if i % 100 == 0 or i == len(unique_elements) - 1:
        print(f"Menerjemahkan baris unik ke-{i} dari {len(unique_elements)}...")

    translations[element] = safe_translate(text_strip)

# Memetakan hasil kembali ke DataFrame utama
df['translated_text'] = df_normalisasi['normalisasi_fixed'].map(translations)

# =========================================================================
# 4. LAPIS 2: POST-TRANSLATE CLEANING & ARTIFACT REMOVAL
# =========================================================================
def post_translate_cleaning(text):
    if not isinstance(text, str):
        return ""

    # Paksa ubah residu kata Indonesia spesifik topik yang gagal ditranslasi
    text = re.sub(r'\bdosen\b', 'lecturer', text, flags=re.IGNORECASE)
    text = re.sub(r'\bgaji\b', 'salary', text, flags=re.IGNORECASE)
    text = re.sub(r'\bahli\b', 'expert', text, flags=re.IGNORECASE)
    text = re.sub(r'\bbayar\b', 'salary', text, flags=re.IGNORECASE)
    text = re.sub(r'\bdpr\b', 'government', text, flags=re.IGNORECASE)

    # Pembersihan karakter aneh dan link
    text = re.sub(r'([^0-9A-Za-z \t])|(\w+:\/\/\S+)', ' ', text)
    text = text.lower()

    # Hapus artefak huruf 's' tunggal sisa apostrof (misal: master's -> master s)
    text = re.sub(r'\b s \b', ' ', text)

    # Menghilangkan spasi ganda
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# Menerapkan Lapis 2
df['translated_text'] = df['translated_text'].apply(post_translate_cleaning)

print("\n✓ Tahap Translation Berhasil Diselesaikan Secara Menyeluruh!")

# Menampilkan hasil akhir
df_display = pd.DataFrame(df[['text', 'cleaning', 'case_folding', 'normalisasi', 'translated_text']])
display(df_display.head(10))
```

```python
synonym_mapping = {
    # a. Kluster Pengajar/Akademisi -> Target Induk: lecturer
    'lecturers': 'lecturer', 'professor': 'lecturer', 'professors': 'lecturer',
    'prof': 'lecturer', 'tutor': 'lecturer', 'tutors': 'lecturer',
    'instructor': 'lecturer', 'instructors': 'lecturer', 'educator': 'lecturer',
    'educators': 'lecturer', 'academic': 'lecturer', 'academics': 'lecturer',
    'academician': 'lecturer', 'faculty': 'lecturer', 'faculties': 'lecturer',
    'teacher': 'lecturer', 'teachers': 'lecturer', 'lecturing': 'lecturer',

    # b. Kluster Finansial/Kesejahteraan -> Target Induk: salary & welfare
    'salaries': 'salary', 'wage': 'salary', 'wages': 'salary',
    'income': 'salary', 'incomes': 'salary', 'pay': 'salary',
    'paid': 'salary', 'paying': 'salary', 'paycheck': 'salary',
    'fee': 'salary', 'honorarium': 'salary', 'honorariums': 'salary',
    'honorary': 'salary', 'honor': 'salary', 'remuneration': 'salary',
    'compensation': 'salary', 'allowance': 'salary', 'allowances': 'salary',
    'stipend': 'salary', 'tukin': 'salary', 'bonus': 'salary',
    'bonuses': 'salary', 'reward': 'salary', 'rewards': 'salary',
    'pension': 'salary', 'pensions': 'salary',
    'prosperity': 'welfare', 'livelihood': 'welfare',

    # c. Kluster Institusi/Tempat Kerja -> Target Induk: institution
    'university': 'institution', 'universities': 'institution', 'college': 'institution',
    'colleges': 'institution', 'campus': 'institution', 'campuses': 'institution',
    'school': 'institution', 'schools': 'institution', 'institutions': 'institution',

    # d. Kluster Sistem & Status Kepegawaian -> Target Induk: system & status
    'application': 'system', 'app': 'system', 'apps': 'system', 'platform': 'system',
    'systems': 'system', 'sister': 'system', 'siakad': 'system', 'sinta': 'system',
    'bkd': 'system', 'dapodik': 'system', 'feeder': 'system',
    'pppk': 'status', 'pns': 'status', 'asn': 'status',
    'tenure': 'status', 'tenured': 'status'
}

def apply_synonyms(text, synonym_dict):
    if pd.isna(text) or not isinstance(text, str) or not text.strip():
        return ""

    # Tangani frasa khusus dulu agar tidak pecah saat di-split
    text = re.sub(r'\bcivil servant\b', 'status', text, flags=re.IGNORECASE)
    text = re.sub(r'\bcivil servants\b', 'status', text, flags=re.IGNORECASE)
    text = re.sub(r'\bnon asn\b', 'status', text, flags=re.IGNORECASE)

    words = text.split()
    replaced = [synonym_dict[w.lower()] if w.lower() in synonym_dict else w for w in words]
    return ' '.join(replaced)

# Jalankan substitusi sinonim pada translated_text hasil cell sebelumnya
df['synonym'] = df['translated_text'].apply(lambda x: apply_synonyms(x, synonym_mapping))
df_display = df[['text', 'cleaning', 'case_folding', 'normalisasi', 'translated_text', 'synonym']]
df_display.head(5)
```

```python
# =========================================================================
# CELL 1: TAHAP STOPWORD REMOVAL (ANTI-NEGASI)
# =========================================================================
import pandas as pd
from nltk.corpus import stopwords

# Mendefinisikan daftar stopword bawaan NLTK
stop_words_id = set(stopwords.words('indonesian'))
stop_words_en = set(stopwords.words('english'))

# PERBAIKAN 1: Kata 'belum' dan 'gak' sudah saya HAPUS dari custom_blacklist ini
custom_blacklist = {
    'even', 'also', 'like', 'really', 'get', 'got', 'would', 'could',
    'make', 'made', 'one', 'two', 'go', 'going', 'say', 'said', 'want',
    'know', 'think', 'well', 'much', 'many', 'way', 'something', 'anything',
    'people', 'still', 'though', 'become', 'yes', 'small', 'year', 'state',
    'first', 'good', 'right', 'take', 'must', 'per', 'sir', 'sis', 'lowk',
    'kalau', 'kita', 'untuk', 'saja', 'ke', 'nya', 'yg', 'aja', 
    'kalo', 'buat', 'dekan', 'mbak', 'ibu', 'pak', 'tapi',
    'yang', 'di', 'ini', 'itu', 'juga', 'kan', 'bagaimana', 'dan', 'harus', 'lagi',
    'sudah', 'dari', 'sama', 'bisa', 'memang', 'dulu', 'bu', 'saya', 'ada',
    'iya', 'begitu', 'terima', 'cuma', 'segini', 'karena', 'baik'
}

# PERBAIKAN 2: Membuat daftar kata negasi yang WAJIB DISELAMATKAN
negation_words = {
    # Kata negasi & slang Bahasa Indonesia
    'tidak', 'bukan', 'belum', 'jangan', 'tak', 'gak', 'enggak', 'ga', 'kurang',
    # Kata negasi Bahasa Inggris (jika ada teks yang sudah terjemahan/campuran)
    'no', 'not', 'nor', 'neither', 'never', 'cannot', 'cant', 'dont', 'doesnt',
    'didnt', 'isnt', 'arent', 'wasnt', 'werent', 'hasnt', 'havent', 'hadnt',
    'wouldnt', 'couldnt', 'shouldnt', 'wont', 'aint'
}

# Gabungkan semua stopword terlebih dahulu
all_stopwords = stop_words_id.union(custom_blacklist).union(stop_words_en)

# PERBAIKAN 3: Kurangi daftar stopword dengan kata negasi.
# Logika: Jika kata negasi ada di all_stopwords, maka akan dihapus dari blacklist.
final_blacklist = all_stopwords - negation_words

# Fungsi pembersihan kata
def final_cleaning_strict(text):
    if pd.isna(text):
        return ""
    words = str(text).split()
    filtered = []

    for w in words:
        w = w.lower()
        # FILTER KETAT: Panjang > 1, bukan angka murni, dan tidak ada di blacklist
        if len(w) > 1 and not w.isdigit() and w not in final_blacklist:
            filtered.append(w)

    return " ".join(filtered)

# Menerapkan pembersihan dan membuat kolom baru
df['stopword_removal'] = df['synonym'].apply(final_cleaning_strict)

# Menampilkan dataframe dengan kolom yang diminta
df[['text', 'cleaning', 'case_folding', 'normalisasi', 'synonym', 'stopword_removal']].head(10)
```

```python
def safe_eval_list(text):
    text = str(text).strip()
    if text.startswith('[') and text.endswith(']'):
        try:
            return ast.literal_eval(text)
        except Exception:
            pass
    # Jika teks berbentuk kalimat biasa (contoh: "lecturer salary")
    text_clean = text.replace("'", "").replace('"', "").replace("[", "").replace("]", "").replace(",", "")
    return text_clean.split()
 
 
# Menerapkan konversi aman ke seluruh baris
texts = df['tokenize'].apply(safe_eval_list).tolist()
 
# Membuat Dictionary (Kamus Kata)
dictionary = Dictionary(texts)
 
# FILTER EXTREMES (Sesuai Paper):
# no_below=3 : Buang kata yang muncul kurang dari 3 kali
# no_above=0.5 : Buang kata yang muncul di lebih dari 50% dokumen
dictionary.filter_extremes(no_below=3, no_above=0.5)
 
# Membuat Corpus (Bag of Words)
corpus = [dictionary.doc2bow(text) for text in texts]
 
 
# =========================================================================
# -> BLOK TAMBAHAN: MENYIMPAN CORPUS & DICTIONARY SECARA PERMANEN
# =========================================================================
print("Membekukan dan menyimpan Dictionary serta Corpus ke folder 'model_sidang'...")
os.makedirs("model_sidang", exist_ok=True)
 
# Simpan Dictionary
dictionary.save("model_sidang/dictionary.dict")
 
# Simpan Corpus menggunakan Pickle
with open("model_sidang/corpus.pkl", "wb") as f:
    pickle.dump(corpus, f)
 
print("✓ Dictionary dan Corpus berhasil dibekukan! Aman untuk fase pemodelan.\n")
 
# =========================================================================
# 2. FUNGSI MENGHITUNG AVERAGE TOPIC OVERLAP / STABILITY (Jaccard Similarity)
# =========================================================================
# Catatan: "overlap" di sini disebut juga "stability score" pada beberapa
# penelitian (mis. Puspitasari et al., 2025, AVITEC Vol.7 No.2,
# DOI: 10.28989/avitec.v7i2.3033), yaitu kemiripan Jaccard antar-topik
# yang dihasilkan model. Nilai lebih rendah = topik lebih berbeda satu
# sama lain (lebih baik).
def calculate_overlap(model, num_words=15):
    topics = []
    for i in range(model.num_topics):
        words = set([word for word, prob in model.show_topic(i, topn=num_words)])
        topics.append(words)
 
    overlaps = []
    for i in range(len(topics)):
        for j in range(i + 1, len(topics)):
            intersection = len(topics[i].intersection(topics[j]))
            union = len(topics[i].union(topics[j]))
            if union > 0:
                overlaps.append(intersection / union)
 
    return np.mean(overlaps) if overlaps else 0
 
 
# =========================================================================
# 3. ITERASI PENCARIAN K OPTIMAL
# =========================================================================
min_topics = 2
max_topics = 10
coherence_scores = []
overlap_scores = []

print("Memulai evaluasi Topic Modeling. Proses ini memakan waktu beberapa menit...")
for k in range(min_topics, max_topics + 1):
    lda_model = LdaModel(
        corpus=corpus,
        id2word=dictionary,
        num_topics=k,
        random_state=56,
        chunksize=50,
        passes=5,
        alpha=0.5,
        per_word_topics=True
    )

    coherence_model = CoherenceModel(model=lda_model, texts=texts, dictionary=dictionary, coherence='c_v')
    c_v = coherence_model.get_coherence()
    coherence_scores.append(c_v)

    overlap = calculate_overlap(lda_model)
    overlap_scores.append(overlap)

    print(f"K={k} | Coherence: {c_v:.4f} | Overlap: {overlap:.4f}")

# =========================================================================
# 4. MENENTUKAN K IDEAL (PENYESUAIAN HEURISTIK UNTUK TESIS)
# =========================================================================
k_values = list(range(min_topics, max_topics + 1))
coherence_arr = np.array(coherence_scores)
overlap_arr = np.array(overlap_scores)

# Selisih (gap) antara coherence dan overlap/stability, nilai mentah
gap_scores = coherence_arr - overlap_arr

# ---------------------------------------------------------
# INTERVENSI PENELITI: Mengunci best_k ke 4
# Berdasarkan analisis semantik dan menjaga agar topik tidak menjadi micro-topics.
# ---------------------------------------------------------
best_k = 6  # <-- DIUBAH DARI OTOMATIS MENJADI MANUAL
best_idx = k_values.index(best_k)

print(f"\n>>> K IDEAL (Heuristik Peneliti) = {best_k} "
      f"(Coherence={coherence_scores[best_idx]:.4f}, "
      f"Overlap={overlap_scores[best_idx]:.4f}, "
      f"Gap={gap_scores[best_idx]:.4f})\n")

print("Rincian gap tiap K:")
for k, c, o, g in zip(k_values, coherence_scores, overlap_scores, gap_scores):
    marker = "  <-- terpilih (heuristik)" if k == best_k else ""
    print(f"K={k} | Coherence={c:.4f} | Overlap={o:.4f} | Gap={g:.4f}{marker}")

# =========================================================================
# 5. PLOT GRAFIK VISUALISASI (DENGAN PENANDA IDEAL TOPIC MANUAL)
# =========================================================================
x = range(min_topics, max_topics + 1)
plt.figure(figsize=(10, 6))

# Plot dua metrik utama
plt.plot(x, overlap_scores, label='Average Topic Overlap (Stability)', marker='o', color='steelblue')
plt.plot(x, coherence_scores, label='Topic Coherence (c_v)', marker='s', color='sandybrown')

# GARIS "IDEAL TOPIC" MANUAL (K=4)
plt.axvline(x=best_k, color='red', linestyle='--', linewidth=2,
            label=f'Ideal Topic (K={best_k})')

plt.xlabel('Number of Topics (K)')
plt.ylabel('Score')
plt.title('Topic Coherence and Stability Overlap')

plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(x)
plt.tight_layout()

os.makedirs("model_sidang", exist_ok=True)
plt.savefig("model_sidang/coherence_overlap_plot_k4.png", dpi=200)
plt.show()
```

```python
import ast
import os
import pickle
import numpy as np
import matplotlib.pyplot as plt
from gensim.corpora import Dictionary
from gensim.models import LdaModel, CoherenceModel
from gensim.models.phrases import Phrases, Phraser

def safe_eval_list(text):
    text = str(text).strip()
    if text.startswith('[') and text.endswith(']'):
        try:
            return ast.literal_eval(text)
        except Exception:
            pass
    text_clean = text.replace("'", "").replace('"', "").replace("[", "").replace("]", "").replace(",", "")
    return text_clean.split()

# 1. Ekstrak teks asli TANPA UBAHAN (Untuk corpus lama/SVM)
texts = df['tokenize'].apply(safe_eval_list).tolist()

# Membuat Dictionary & Corpus LAMA (Sama persis sebagai pembanding murni)
dictionary_lama = Dictionary(texts)
dictionary_lama.filter_extremes(no_below=3, no_above=0.5)
corpus_lama = [dictionary_lama.doc2bow(text) for text in texts]

# =========================================================================
# -> BLOK TAMBAHAN: ENTITY RECOGNITION (BIGRAM) & EXTENDED STOPWORDS
# =========================================================================
# A. Perluasan Custom Stopwords
custom_stopwords = {
    'no', 'not', 'noy', 
    'come', 'look', 'try', 'let', 'already', 'actually', 'whose', 'different',
    'mrs', 'oh', 'regarding', 'considered', 'indeed', 'etc',
    'wow', 'yesterday', 'especially', 'looking', 'called', 'big', 'high'
}

# B. Manual Entity Recognition (Menggabungkan kata spesifik)
def apply_manual_entities(tokens):
    text_joined = " ".join(tokens)
    # Tambahkan entitas lain di bawah ini sesuai kebutuhan riset Anda
    text_joined = text_joined.replace("sri mulyani", "sri_mulyani")
    text_joined = text_joined.replace("prabowo subianto", "prabowo_subianto")
    text_joined = text_joined.replace("bpjs ketenagakerjaan", "bpjs_ketenagakerjaan")
    text_joined = text_joined.replace("non asn", "non_asn")
    return text_joined.split()

texts_manual_entities = [apply_manual_entities(t) for t in texts]

# C. Otomatisasi Bigram dengan Gensim (Menangkap pasangan kata lain yang sering berdampingan)
phrases = Phrases(texts_manual_entities, min_count=3, threshold=10)
bigram_model = Phraser(phrases)
texts_bigramed = [bigram_model[t] for t in texts_manual_entities]

# D. Hapus Custom Stopwords dari teks yang sudah di-Bigram
texts_lda = [[word for word in text if word.lower() not in custom_stopwords] for text in texts_bigramed]

# Membuat Dictionary & Corpus KHUSUS LDA BARU
dictionary_lda = Dictionary(texts_lda)
dictionary_lda.filter_extremes(no_below=3, no_above=0.5)
corpus_lda = [dictionary_lda.doc2bow(text) for text in texts_lda]

# =========================================================================
# -> MENYIMPAN CORPUS & DICTIONARY SECARA PERMANEN
# =========================================================================
print("Membekukan dan menyimpan Dictionary serta Corpus ke folder 'model_sidang'...")
os.makedirs("model_sidang", exist_ok=True)

dictionary_lama.save("model_sidang/dictionary_lama.dict")
with open("model_sidang/corpus_lama.pkl", "wb") as f:
    pickle.dump(corpus_lama, f)

dictionary_lda.save("model_sidang/dictionary_lda_v2.dict")
with open("model_sidang/corpus_lda_v2.pkl", "wb") as f:
    pickle.dump(corpus_lda, f)

print("✓ Dictionary dan Corpus (Lama & LDA V2) berhasil dibekukan!\n")

# =========================================================================
# 2. FUNGSI MENGHITUNG AVERAGE TOPIC OVERLAP / STABILITY (Jaccard Similarity)
# =========================================================================
def calculate_overlap(model, num_words=15):
    topics = []
    for i in range(model.num_topics):
        words = set([word for word, prob in model.show_topic(i, topn=num_words)])
        topics.append(words)

    overlaps = []
    for i in range(len(topics)):
        for j in range(i + 1, len(topics)):
            intersection = len(topics[i].intersection(topics[j]))
            union = len(topics[i].union(topics[j]))
            if union > 0:
                overlaps.append(intersection / union)
    return np.mean(overlaps) if overlaps else 0

# =========================================================================
# 3. ITERASI PENCARIAN K OPTIMAL
# =========================================================================
min_topics = 2
max_topics = 10
coherence_scores = []
overlap_scores = []

print("Memulai evaluasi Topic Modeling (menggunakan Corpus LDA V2 - Bigram & Cleaned)...")
for k in range(min_topics, max_topics + 1):
    lda_model = LdaModel(
        corpus=corpus_lda,
        id2word=dictionary_lda,
        num_topics=k,
        random_state=56,
        chunksize=50,
        passes=5,
        alpha=0.5,
        per_word_topics=True
    )

    coherence_model = CoherenceModel(
        model=lda_model, 
        texts=texts_lda, 
        dictionary=dictionary_lda, 
        coherence='c_v'
    )
    c_v = coherence_model.get_coherence()
    coherence_scores.append(c_v)

    overlap = calculate_overlap(lda_model)
    overlap_scores.append(overlap)

    print(f"K={k} | Coherence: {c_v:.4f} | Overlap: {overlap:.4f}")

# =========================================================================
# 4. MENENTUKAN K IDEAL 
# =========================================================================
k_values = list(range(min_topics, max_topics + 1))
coherence_arr = np.array(coherence_scores)
overlap_arr = np.array(overlap_scores)
gap_scores = coherence_arr - overlap_arr

# K-Ideal kita set berdasarkan analisis Anda nanti. Sementara di-set 4.
best_k = 6
best_idx = k_values.index(best_k)

print(f"\n>>> K IDEAL (Manual) = {best_k} "
      f"(Coherence={coherence_scores[best_idx]:.4f}, "
      f"Overlap={overlap_scores[best_idx]:.4f}, "
      f"Gap={gap_scores[best_idx]:.4f})\n")

print("Rincian gap tiap K:")
for k, c, o, g in zip(k_values, coherence_scores, overlap_scores, gap_scores):
    marker = "  <-- terpilih" if k == best_k else ""
    print(f"K={k} | Coherence={c:.4f} | Overlap={o:.4f} | Gap={g:.4f}{marker}")

# =========================================================================
# 5. PLOT GRAFIK VISUALISASI 
# =========================================================================
x = range(min_topics, max_topics + 1)
plt.figure(figsize=(10, 6))

plt.plot(x, overlap_scores, label='Average Topic Overlap (Stability)', marker='o', color='steelblue')
plt.plot(x, coherence_scores, label='Topic Coherence (c_v)', marker='s', color='sandybrown')

plt.axvline(x=best_k, color='red', linestyle='--', linewidth=2, label=f'Ideal Topic (K={best_k})')

plt.xlabel('Number of Topics (K)')
plt.ylabel('Score')
plt.title('Topic Coherence and Stability Overlap (LDA V2 - Bigram)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(x)
plt.tight_layout()

os.makedirs("model_sidang", exist_ok=True)
plt.savefig(f"model_sidang/coherence_overlap_plot_k{best_k}_v2.png", dpi=200) 
plt.show()
```

```python
# =========================================================================
# 0. MEMBUAT FINAL MODEL LDA (K = 6)
# =========================================================================
print("Membangun Final Model LDA dengan K=6...")
final_lda_model = LdaModel(
    corpus=corpus_lda,
    id2word=dictionary_lda,
    num_topics=6,       # K Optimal Anda
    random_state=56,    # Sesuai Paper
    chunksize=50,       # Sesuai Paper
    passes=5,           # Sesuai Paper
    alpha=0.5,          # Sesuai Paper
    per_word_topics=True
)

# =========================================================================
# 1. FUNGSI EKSTRAKSI TOPIK DOMINAN (ANTI-ERROR)
# =========================================================================
def format_topics_sentences(ldamodel, corpus_data):
    dominant_topics = []

    # Looping melalui setiap dokumen di dalam corpus
    for i, row_tuple in enumerate(ldamodel[corpus_data]):
        # Karena per_word_topics=True, ekstrak indeks ke-0 (doc_topics) saja
        doc_topics = row_tuple[0]

        # Mengurutkan probabilitas topik dari yang paling besar ke kecil
        doc_topics_sorted = sorted(doc_topics, key=lambda x: x[1], reverse=True)

        # Mengambil ID topik urutan pertama (paling dominan)
        topic_num = doc_topics_sorted[0][0]
        dominant_topics.append(int(topic_num))

    return dominant_topics

# Memasukkan Topik ID ke dalam dataframe utama df_lda
df['Topic_ID'] = format_topics_sentences(ldamodel=final_lda_model, corpus_data=corpus_lda)

# =========================================================================
# 2. MAPPING TOPIK ID KE ASPECT NAME & CETAK TABEL FREKUENSI
# =========================================================================
# Pastikan urutan nama ini sesuai dengan output keywords yang kita bahas sebelumnya
aspect_names = {
    0: "Government Welfare Policy",
    1: "Salary & Academic Qualification",
    2: "National Education Budget",
    3: "Institutional Burden & State Fund Allocation",
    4: "Non-ASN Employment Status",
    5: "General Education Discourse",
}

df['Aspect_Name'] = df['Topic_ID'].map(aspect_names)

aspek_frekuensi = df['Aspect_Name'].value_counts().reset_index()
aspek_frekuensi.columns = ['Nama Aspek', 'Frekuensi Dokumen']

print("\n=== DISTRIBUSI FREKUENSI ASPEK ===")
display(aspek_frekuensi)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.feature_extraction.text import TfidfVectorizer, ENGLISH_STOP_WORDS
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split, cross_val_predict, StratifiedKFold
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

import warnings
from sklearn.exceptions import ConvergenceWarning

# Menekan peringatan untuk tampilan output yang lebih bersih
warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", category=ConvergenceWarning)

# =========================================================================
# 1. PERSIAPAN DATA (PENDEKATAN 2 KELAS SENTIMEN)
# =========================================================================
# Filter ketat hanya untuk Positif dan Negatif
df_model = df[df['Sentiment'].isin(['Positif', 'Negatif'])].copy()

# Fitur X (Baseline) dan Target Y (Sentimen)
X = df_model['stopword_removal'] 
y_label = df_model['Sentiment']

print(f"Total dokumen diproses (Biner): {len(df_model)} (Positif/Negatif)")

# =========================================================================
# 2. TERM WEIGHTING (TF-IDF ANTI-NEGASI - UNIGRAM)
# =========================================================================
# Menyusun kamus stopword empiris: Hapus kata negasi dari daftar buang
standard_stopwords = set(ENGLISH_STOP_WORDS)
negation_words = {'not', 'no', 'nor', 'neither', 'never', 'cannot', 'cant', 'dont', 'doesnt'}
custom_stopwords = list(standard_stopwords - negation_words)

# Terapkan custom_stopwords dengan N-gram (1,1) sesuai hasil eksperimen terbaik
tfidf_vectorizer = TfidfVectorizer(
    ngram_range=(1, 1),
    stop_words=custom_stopwords 
)
X_tfidf = tfidf_vectorizer.fit_transform(X)

# =========================================================================
# 3. SETUP MODEL SVM
# =========================================================================
svm_model = SVC(
    kernel='linear',
    C=1.0,
    class_weight='balanced',
    probability=True, 
    random_state=42,
    max_iter=-1
)

print("✅ Persiapan data Baseline, ekstraksi TF-IDF (Unigram), dan Setup Model SVM selesai.")
print(f"Dimensi matriks TF-IDF: {X_tfidf.shape}")
```
